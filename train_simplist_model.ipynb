```python
yaml_path = '/content/dataset_formes/data.yaml'

yaml_content = """
path: /content/dataset_formes/dataset_formes
train: images/train
val: images/train
nc: 3
names: ['cercle', 'carre', 'triangle']
"""
with open("/content/dataset_formes/data.yaml", "w") as f:
    f.write(yaml_content)
print("Fichier YAML mis à jour pour Colab !")
```

```python
model = YOLO('yolov8n.pt')

results = model.train(
    data='/content/dataset_formes/data.yaml',
    epochs=15,
    imgsz=640,
    batch=16,
    device=0,
    project='runs',
    name='entrainement_formes'
)
```
